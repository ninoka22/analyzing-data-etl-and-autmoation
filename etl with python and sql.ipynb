```python
import pandas as pd
import os

files = {
    "01_خدمات_فناوری_خرداد": r"E:\hasin\01_خدمات_فناوری_خرداد_1405.xlsx",
    "02_پشتیبانی_خرداد": r"E:\hasin\02_پشتیبانی_خرداد_1405.xlsx",
    "03_تعداد_نفرات_خرداد": r"E:\hasin\03_تعداد_نفرات_خرداد_1405.xlsx",
    "04_خدمات_فناوری_تیر": r"E:\hasin\04_خدمات_فناوری_تیر_1405.xlsx",
    "05_پشتیبانی_تیر": r"E:\hasin\05_پشتیبانی_تیر_1405.xlsx",
    "06_تعداد_نفرات_تیر": r"E:\hasin\06_تعداد_نفرات_تیر_1405.xlsx"
}

def profile_file(file_name, file_path):

    print("\n" + "=" * 90)
    print(f"FILE: {file_name}")
    print("=" * 90)

    # Read Excel
    df = pd.read_excel(file_path)

    print("\n--- BASIC INFORMATION ---")

    print(f"File path      : {file_path}")
    print(f"Rows           : {df.shape[0]}")
    print(f"Columns        : {df.shape[1]}")
    print(f"Memory usage   : {df.memory_usage(deep=True).sum() / 1024:.2f} KB")

    print("\n--- COLUMN INFORMATION ---")

    column_info = pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str).values,
        "Non-Null": df.notna().sum().values,
        "Null": df.isna().sum().values,
        "Unique": df.nunique(dropna=True).values
    })

    print(column_info.to_string(index=False))

    print("\n--- DUPLICATES ---")

    duplicate_count = df.duplicated().sum()

    print(f"Duplicate rows : {duplicate_count}")

    print("\n--- MISSING VALUES ---")

    missing = df.isna().sum()
    missing = missing[missing > 0]

    if len(missing) == 0:
        print("No missing values.")
    else:
        print(missing.to_string())

    print("\n--- FIRST 5 ROWS ---")

    print(df.head().to_string(index=False))

    print("\n--- UNIQUE VALUES ---")

    for col in df.columns:

        unique_count = df[col].nunique(dropna=True)

        # Only display columns with relatively few unique values
        if unique_count <= 20:

            print(f"\n[{col}]")
            print(df[col].dropna().unique())

    return df

dataframes = {}

for file_name, file_path in files.items():

    if os.path.exists(file_path):

        dataframes[file_name] = profile_file(
            file_name,
            file_path
        )

    else:

        print("\n" + "!" * 90)
        print(f"FILE NOT FOUND: {file_path}")
        print("!" * 90)
```

```python
def fa(text):
    if not isinstance(text, str):
        return text
    return get_display(arabic_reshaper.reshape(text))

for key, df in raw_frames.items():
    print("=" * 60, key)

    # نمودارهای ستون‌های دسته‌ای
    for c in df.columns:
        if df[c].dtype == object and df[c].nunique() <= 12:
            counts = df[c].value_counts()
            labels = [fa(x) for x in counts.index]

            ax = counts.plot(kind="bar")
            ax.set_xticks(range(len(labels)))
            ax.set_xticklabels(labels, rotation=45, ha="right")
            ax.set_title(fa(f"{key} - {c}"))
            ax.set_xlabel(fa(c))
            ax.set_ylabel(fa("تعداد"))

            plt.tight_layout()
            plt.savefig(os.path.join(CHARTS, f"{key}_{c}_bar.png"))
            plt.show()

    # نمودارهای هیستوگرام ستون‌های عددی
    num_cols = df.select_dtypes(include="number").columns
    for c in num_cols:
        ax = df[c].dropna().plot(kind="hist", bins=8)
        ax.set_title(fa(f"{key} - توزیع {c}"))
        ax.set_xlabel(fa(c))
        ax.set_ylabel(fa("تعداد"))
        ax.ticklabel_format(style="plain", axis="x")

        plt.tight_layout()
        plt.savefig(os.path.join(CHARTS, f"{key}_{c}_hist.png"))
        plt.show()
```

```python
def standardize(df, key, category, period):
    df = df.rename(columns=COLUMN_MAPPING[key])
    df["cost_category"] = category
    df["source_file"]   = os.path.basename(FILES_TIR[key]["path"])

    if "period" not in df.columns:
        df["period"] = period
    else:
        df["period"] = df["period"].apply(norm_period)

    if "company_name" in df.columns:
        df["company_name_raw"] = df["company_name"].apply(norm_text)
        df["company_name"]     = df["company_name_raw"].map(COMPANY_MAPPING)
        df["company_unknown"]  = df["company_name"].isna()
    else:
        df["company_name"] = None
        df["company_unknown"] = True

    if "cost_title" in df.columns:
        df["cost_title"] = df["cost_title"].apply(norm_text)

    if "amount" in df.columns:
        df["amount_raw"] = df["amount"]
        df["amount"]     = df["amount"].apply(clean_amount)
        df["amount_missing"]  = df["amount"].isna()
        df["amount_negative"] = df["amount"].fillna(0) < 0
    else:
        df["amount"] = None
        df["amount_missing"] = True
        df["amount_negative"] = False

    if "allocation_basis" in df.columns:
        df["allocation_basis_raw"] = df["allocation_basis"].apply(norm_text)
        df["allocation_basis"]     = df["allocation_basis_raw"].map(ALLOCATION_MAPPING)
    else:
        df["allocation_basis"] = None

    if "headcount" in df.columns:
        df["headcount_raw"]     = df["headcount"]
        df["headcount"]         = df["headcount"].apply(clean_amount)
        df["headcount_missing"] = df["headcount"].isna()
    else:
        df["headcount_missing"] = False

    return df
```

```python
q = """
SELECT
  period,
  COUNT(DISTINCT company_name) AS companies,
  SUM(total_allocated)         AS total_allocated,
  SUM(direct_amount)           AS total_direct,
  SUM(equal_amount)            AS total_equal,
  SUM(headcount_amount)        AS total_headcount,
  AVG(total_allocated)         AS avg_per_company,
  MAX(total_allocated)         AS max_company_amount,
  MIN(total_allocated)         AS min_company_amount
FROM fact_company_period
GROUP BY period
"""
print("شاخص‌های کلان:")
print(pd.read_sql(q, conn))
```

```python
q = """
SELECT
  k.company_name,
  k.total_allocated AS khordad_amount,
  t.total_allocated AS tir_amount,
  ROUND((t.total_allocated - k.total_allocated) * 100.0 /
        k.total_allocated, 2) AS change_percent
FROM fact_company_period k
JOIN fact_company_period t
  ON t.company_name = k.company_name AND t.period = '1405/04'
WHERE k.period = '1405/03'
  AND ABS((t.total_allocated - k.total_allocated) * 100.0 /
          k.total_allocated) > 30
ORDER BY ABS(change_percent) DESC
"""
print("موارد غیرعادی (تغییر بیش از سی درصد):")
print(pd.read_sql(q, conn))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import arabic_reshaper
from bidi.algorithm import get_display

plt.rcParams["font.family"] = "Tahoma"
plt.rcParams["axes.unicode_minus"] = False
sns.set_style("whitegrid")

def fa(text):
    if not isinstance(text, str):
        return text
    return get_display(arabic_reshaper.reshape(text))

pivot = fact_company.pivot(index="company_name", columns="period",
                            values="total_allocated").fillna(0)
pivot = pivot.drop("همه شرکت ها", errors="ignore")

xlabels = [fa(x) for x in pivot.index]
legend_labels = [fa(str(x)) for x in pivot.columns]

ax = pivot.plot(kind="bar")
ax.set_xticks(range(len(xlabels)))
ax.set_xticklabels(xlabels, rotation=45, ha="right")
ax.set_title(fa("مقایسه تخصیص هر شرکت در خرداد و تیر"))
ax.set_xlabel(fa("نام شرکت"))
ax.set_ylabel(fa("مبلغ"))
ax.ticklabel_format(style="plain", axis="y")
ax.legend(title=fa("دوره"), labels=legend_labels, loc="best")

plt.tight_layout()
plt.savefig(r"E:\hasin\charts\final_company_compare.png")
plt.show()
```
